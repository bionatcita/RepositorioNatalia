Una cepa de E. Coli tiene una tasa de crecimiento tal que un individuo se duplica cada 30 minutos.
Calcula en cuánto tiempo esta cepa
a) Llenaría el volumen de la Tierra (considerandola como una esfera hueca)
b) Cubriría la superficie de la Tierra
c) Repite el cálculo para una tasa de crecimiento:
   i)se duplica cada hora
   ii)se duplica 
   

```python
import math as mt
```

```python
pi = 3.1416
rT = 6371e3
rE = 0.5e-6
```

VT = 4/3 (pi) (rT)**3
VE = 4/3 (PI) (rE) **3

N= cuantas e. coli llenarían el volumen de la tierra?
N= VT/VE
N= (4/3*pi*rT**3)/(4/3*pi*rE**3)
se elimina "4/3*pi" porque está igual a ambos lados de la división 


```python
N = (6371e3)**3/(0.5e-6)**3
```
